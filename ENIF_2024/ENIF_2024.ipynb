##### 8. Análisis de resultados #####

```python
# Población adulta total representada
poblacion_total = df["FAC_PER"].sum()

# Población representada con cuenta formal
poblacion_cuenta = df.loc[df["cuenta_formal"] == 1, "FAC_PER"].sum()

# Porcentaje ponderado
porcentaje_cuenta = (poblacion_cuenta / poblacion_total) * 100

print("Población adulta representada:", poblacion_total)
print("Población con cuenta formal:", poblacion_cuenta)
print("Porcentaje con cuenta formal:", round(porcentaje_cuenta, 2), "%")
```

El 63.26% de la población adulta representada por la ENIF 2024 tiene una cuenta formal, según la definición que construimos con P5_4_1 a P5_4_7.

```python
# Porcentaje con cuenta formal según sexo

resultado_sexo = (
    df.groupby("sexo_cat")
    .apply(
        lambda x: (
            x.loc[x["cuenta_formal"] == 1, "FAC_PER"].sum()
            / x["FAC_PER"].sum()
        ) * 100
    )
)

resultado_sexo.round(2)
```

```python
# Porcentaje con cuenta formal según nivel educativo

resultado_educacion = (
    df.groupby("nivel_educativo")
    .apply(
        lambda x: (
            x.loc[x["cuenta_formal"] == 1, "FAC_PER"].sum()
            / x["FAC_PER"].sum()
        ) * 100
    )
)

resultado_educacion.round(2).sort_values(ascending=False)
```

```python
# Porcentaje con cuenta formal según tamaño de localidad

resultado_localidad = (
    df.groupby("tam_localidad")
    .apply(
        lambda x: (
            x.loc[x["cuenta_formal"] == 1, "FAC_PER"].sum()
            / x["FAC_PER"].sum()
        ) * 100
    )
)

resultado_localidad.round(2)
```

el porcentaje de adultos con cuenta formal disminuye conforme disminuye el tamaño de la localidad.

```python
##### 9. Visualización de resultados #####

# Cuenta formal según sexo
resultado_sexo.plot(kind="bar")

plt.title("Población con cuenta formal según sexo")
plt.xlabel("Sexo")
plt.ylabel("Porcentaje (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)

plt.show()
```
