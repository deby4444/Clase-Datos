### Pregunta 13

¿Cuantos socios tienen una edad fuera de un rango logico, es decir, menor a 16 o mayor a 75 anios?

````python
# Tu codigo aqui
((df["edad"] < 16) | (df["edad"] > 75)).sum()
````

### Pregunta 14

Usando el criterio del rango intercuartilico (IQR = Q3 - Q1; un valor es outlier si es menor a `Q1 - 1.5*IQR` o mayor a `Q3 + 1.5*IQR`), ¿cuantos valores atipicos se detectan en la columna `calorias_quemadas`?

````python
# Tu codigo aqui
Q1 = df["calorias_quemadas"].quantile(0.25)
Q3 = df["calorias_quemadas"].quantile(0.75)

IQR = Q3 - Q1

lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

((df["calorias_quemadas"] < lim_inf) |
 (df["calorias_quemadas"] > lim_sup)).sum()
````

## 7. Tabla cruzada: nivel de actividad por rango de edad

### Pregunta 15

Elimina de `df` las filas con edad fuera del rango logico (menor a 16 o mayor a 75; mismo criterio de la Pregunta 13) y sobrescribe `df`. Crea una nueva columna `rango_edad` con:

```python
pd.cut(df["edad"], bins=[16, 30, 45, 60, 75],
       labels=["16-30", "31-45", "46-60", "61-75"], include_lowest=True)
```

Construye una tabla cruzada con `pd.crosstab(df["nivel_actividad"], df["rango_edad"])`. Segun esa tabla, ¿cuantos socios con `nivel_actividad` **"Activo"** estan en el rango de edad **"16-30"**?

````python
# Tu codigo aqui
#Eliminar a los 4 socios
df = df[(df["edad"] >= 16) & (df["edad"] <= 75)].reset_index(drop=True)
````

````python
#verificar
((df["edad"] < 16) | (df["edad"] > 75)).sum()
````

````python
#crear el rango de edad correcto
df["rango_edad"] = pd.cut(
    df["edad"],
    bins=[16, 30, 45, 60, 75],
    labels=["16-30", "31-45", "46-60", "61-75"],
    include_lowest=True
)
````

````python
#tabla cruzada
tabla = pd.crosstab(df["nivel_actividad"], df["rango_edad"])

tabla
````
